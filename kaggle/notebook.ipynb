# NFT Press Release FAQ Generator: Search Walkthrough

This notebook runs the [NFT Press Release FAQ Generator](https://github.com/buycryptopressrelease/nft-press-release-faq-generator) on 50 real world questions about NFT press releases, checks each one finds the expected FAQ, and generates the FAQ with FAQPage schema. Source: [NFT Project PR](https://buycryptopressrelease.com/nft-project-pr/).

## 1. Load the tool and the datasets

```python
import os, sys
import pandas as pd

BASES = ["/kaggle/input/nft-press-release-faq-dataset", "..", "."]
base = next(b for b in BASES if os.path.exists(os.path.join(b, "nft_pr_faq.py")))
data = base if os.path.exists(os.path.join(base, "nft_pr_faqs.csv")) else os.path.join(base, "dataset")
sys.path.insert(0, base)
import nft_pr_faq as nf

faqs = pd.read_csv(os.path.join(data, "nft_pr_faqs.csv"))
bench = pd.read_csv(os.path.join(data, "nft_pr_search_benchmarks.csv"))
print(len(faqs), "FAQs |", len(bench), "benchmark questions")
```

## 2. FAQs per category

```python
faqs.groupby("category").size().rename("faqs").to_frame()
```

## 3. Run every benchmark question

```python
rows = []
for r in bench.itertuples():
    a = nf.ask(r.query)
    rows.append({"query": r.query, "found": a["id"], "expected": r.expected_id,
                 "confidence": a["confidence"], "band": a["band"], "pass": a["id"] == r.expected_id})
results = pd.DataFrame(rows)
print(f"{results['pass'].sum()}/{len(results)} benchmark questions matched")
results
```

```python
results.groupby("band").size().rename("questions").to_frame()
```

## 4. Ask your own question

```python
for r in nf.search("how early should i publish before my mint", 3):
    print(f"Q{r['id']} [{r['confidence']}] {r['question']}")

best = nf.ask("how much does an nft press release cost")
print("\n", best["answer"])
```

## 5. Generate the FAQ with FAQPage schema

```python
print(nf.export("jsonld", "launch")[:1200])
```
